# Python AES 解密

```python
import base64
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import padding

def decrypt_aes(encrypted_text, key_str, iv_str):
    ct = base64.b64decode(encrypted_text)
    key_bytes = key_str.encode('utf-8')
    iv_bytes = iv_str.encode('utf-8')
    cipher = Cipher(algorithms.AES(key_bytes), modes.CBC(iv_bytes), backend=default_backend())
    decryptor = cipher.decryptor()
    padded_data = decryptor.update(ct) + decryptor.finalize()
    unpadder = padding.PKCS7(128).unpadder()
    plain_text = unpadder.update(padded_data) + unpadder.finalize()
    return plain_text.decode('utf-8')

# --- 互動輸入區 ---
print("--- 進入解密模式 ---")
input_encrypted = input("1. 請輸入『加密後的結果』: ").strip()
input_key = input("2. 請輸入『Key (32字元)』: ").strip()
input_iv = input("3. 請輸入『IV (16字元)』: ").strip()

try:
    decrypted_password = decrypt_aes(input_encrypted, input_key, input_iv)
    print(f"\n--------------------------------------")
    print(f"✅ 解密成功！")
    print(f"原始內容為: {decrypted_password}")
    print(f"--------------------------------------")
except Exception as e:
    print(f"\n❌ 解密失敗！請檢查輸入的資訊是否正確。")
    print(f"錯誤原因: {e}")
```

# 綜合版功能

```python
import secrets
import string
import base64
import getpass  # 用於隱藏密碼輸入
from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import padding

# --- 功能函式 ---

def encrypt_aes(plain_text, key_str, iv_str):
    key_bytes = key_str.encode('utf-8')
    iv_bytes = iv_str.encode('utf-8')
    padder = padding.PKCS7(128).padder()
    padded_data = padder.update(plain_text.encode('utf-8')) + padder.finalize()
    cipher = Cipher(algorithms.AES(key_bytes), modes.CBC(iv_bytes), backend=default_backend())
    encryptor = cipher.encryptor()
    ct = encryptor.update(padded_data) + encryptor.finalize()
    return base64.b64encode(ct).decode('utf-8')

def decrypt_aes(encrypted_text, key_str, iv_str):
    ct = base64.b64decode(encrypted_text)
    key_bytes = key_str.encode('utf-8')
    iv_bytes = iv_str.encode('utf-8')
    cipher = Cipher(algorithms.AES(key_bytes), modes.CBC(iv_bytes), backend=default_backend())
    decryptor = cipher.decryptor()
    padded_data = decryptor.update(ct) + decryptor.finalize()
    unpadder = padding.PKCS7(128).unpadder()
    plain_text = unpadder.update(padded_data) + unpadder.finalize()
    return plain_text.decode('utf-8')

def generate_random_string(length):
    alphabet = string.ascii_letters + string.digits
    return ''.join(secrets.choice(alphabet) for _ in range(length))

# --- 互動介面 ---

print("=== AES 加解密互動工具 ===")
mode = input("請選擇模式 (1: 加密 / 2: 解密): ")

if mode == '1':
    # 使用 getpass 隱藏密碼輸入，畫面不會顯示你打了什麼
    target_pw = getpass.getpass("請輸入要加密的密碼: ")

    # 自動產生 Key 與 IV
    new_key = generate_random_string(32)
    new_iv = generate_random_string(16)

    result = encrypt_aes(target_pw, new_key, new_iv)

    print("\n--- 加密結果 ---")
    print(f"Key (32字元): {new_key}")
    print(f"IV  (16字元): {new_iv}")
    print(f"加密後的密碼: {result}")
    print("---------------")

elif mode == '2':
    input_ct = input("請貼上『加密後的密碼』: ")
    input_key = input("請輸入 Key (32字元): ")
    input_iv = input("請輸入 IV (16字元): ")

    try:
        original_pw = decrypt_aes(input_ct, input_key, input_iv)
        print(f"\n✅ 解密成功！原始密碼為: {original_pw}")
    except Exception as e:
        print(f"\n❌ 解密失敗，請檢查 Key/IV 或加密字串是否正確。")
        print(f"錯誤訊息: {e}")

else:
    print("無效的選擇，請重新執行程式。")
```
